#Banking Application -> ATM Machine

```python
class AtmMachine:
    def __init__(self):
        self.pin = ""
        self.balance = 0
        self.menu()  #this will call it self without the user manually typing AtmMachine.menu()

    def menu(self):
        user_input = input(
            """
            Hi how i can help you?

            1. Press 1 to create pin
            2. Press 2 to change pin
            3. Press 3 to check balence 
            4. Press 4 to withdraw 
            5. Anything to exit

            """)

        match user_input:
            case "1":
                self.create_pin()
            case "2":
                pass
            case "3":
                pass
            case "4":
                pass
            case "5":
                pass
            case _:
                print("Invalid Choice")
                exit()

    def create_pin(self):
        user_pin = input("Enter your pin: ")
        self.pin = user_pin

        user_balance = int(input("Enter Balance: "))
        self.balance = user_balance

        print("Pin Created Successfully")
        self.menu()
```
