`````python
class Person:
    def __init__(self, name, country):
        self.name = name           # This is how we will save data in side the constructor.
        self.country = country 
    def greeting(self):
        if self.country.lower() == "india":
            print(f"Namaste {self.name}")
        else:
            print(f"Hello {self.name}")
`````

# Pass by reference in OOP
If the function takes class object as an attribute.

`````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Defining a function 
def greating(Person):
    print(f"Hello i am {Person.name} and I am {Person.age} year old.")

p = Person("Sumit", 12)

greating(p)
`````

**You can also return a one specific class function as an object**

`````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Defining a function 
def greeting(person):
    print(f"Hello i am {person.name} and I am {person.age} year old.")
    p1 = Person("Alex", 23)
    return p1

`````

`````python
p = Person("baby", 25)
`````

`````python
a = greeting(p)
`````

`````python
a.name
`````

`````python
a.age
`````

As we can see even if we have changed the name and age and use the greeting function then too we got the name and age of the class object.

# Encapsulation
It protects classes form accidental changes or deletions and promotes code reusability and maintainability.

when ever you will use "__" in front of a variable it will become a private property or can also say class variable. Which means the variable can not be changed from outside of the class.

`````python
class Car:
    def __init__(self):
        self.color = "Red"
        self.__wheels = 4

    def show_wheels(self):
        car_color = self.color
        car_wheels = self.__wheels
        print(f"Car color is {car_color} and no. of wheels are {car_wheels}")
`````

`````python
obj = Car()
`````

`````python
obj.color
`````

`````python
obj.color = "Blue"
`````

`````python
obj.color
`````

But in the case of private variable it will not change. And you can not even access the private variable normally try running the follwoing code and observe the error.

`````python
# obj.__wheels
`````

`````python
obj.__wheels = 6
`````

`````python
obj.show_wheels()
`````

You can see the modified changes are not reflecting to the private variable.

But you can still access it. Python changes the name internally form ```__wheels``` to ```_Car__wheels```

`````python
obj._Car__wheels
`````

but through a function it can be access 

`````python
obj._Car__wheels = 5
`````
