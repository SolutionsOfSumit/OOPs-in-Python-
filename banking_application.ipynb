# How object access attributes in OOP
Just like in function we can also give attributes in side the class as well by just writing it after the self attribute.

`````python
class Person:
    def __init__(self, name, country):
        self.name = name           # This is how we will save data in side the constructor.
        self.country = country 
    def greeting(self):
        if self.country.lower() == "india":
            print(f"Namaste {self.name}")
        else:
            print(f"Hello {self.name}")
`````

`````python
# Defining the object of the above class
p = Person("Sumit", "India")
`````

`````python
p.greeting()
`````

`````python
p.name
`````

# Pass by reference in OOP
If the function takes class object as an attribute.

`````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Defining a function 
def greating(Person):
    print(f"Hello i am {Person.name} and I am {Person.age} year old.")

p = Person("Sumit", 12)

greating(p)
`````

**You can also return a one specific class function as an object**

`````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Defining a function 
def greeting(Person):
    print(f"Hello i am {Person.name} and I am {Person.age} year old.")
    p1 = Person("Alex", 23)
    return p1

`````
