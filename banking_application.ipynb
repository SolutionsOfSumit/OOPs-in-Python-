# Encapsulation
It protects classes form accidental changes or deletions and promotes code reusability and maintainability.

when ever you will use "__" in front of a variable it will become a private property or can also say class variable. Which means the variable can not be changed from outside of the class.

### Setter and Getter Method

##### Getter Method
is a method that retrieves the value of private or protected attribute, instead of letting other code access directly. 

`````python
class WildAnimals:
    def __init__(self):
        self.color = ""
        self.__legs = 4 # this is a private property
    
    def get_legs(self):
        return self.__legs
`````

`````python
obj = WildAnimals()
`````

`````python
obj.get_legs()
`````

##### Setter Method
Now lets say you want to change that same private variable then we have to take help of setter method.

`````python
class WildAnimal:
    def __init__(self):
        self.color = ""
        self.__legs = 4 # this is a private property
    
    def get_legs(self):
        return self.__legs
    
    def set_legs(self, new_legs):
        if new_legs == int:
            self.__legs = new_legs
        else:
            print("give a number")
`````

`````python
obj = WildAnimal()
obj.set_legs(6)
`````

`````python
obj.get_legs()
`````

# Static Variable and Methods inside OOPs
Static variable is a better way to use instead of instance variable which changes itself everytime the the instance is created.

for example

`````python
class Customers:
    def __init__(self):
        self.name = ""
        self.customer_id = 0
        self.customer_id += 1
`````

Acrding to the above code everytime a customer will be created the customer_id will increment itself.

`````python
c1 = Customers()
c1.customer_id
`````

`````python
c2 = Customers()
c2.customer_id
`````

In both cases it is giving us the same id which is not good for our project for that we use **Static variables**.

## Static Variables
are defined inside the class not inside the constructor. Which is accessed by the class name even inside the same class.

`````python
class Customers:

    customer_id = 0

    def __init__(self):
        self.name = ""
        self.customer_id = Customers.customer_id
        Customers.customer_id += 1
`````

`````python
c1 = Customers()
c1.customer_id
`````

`````python
Customers.customer_id
`````

`````python
c2 = Customers()
Customers.customer_id
`````

`````python
c2 = Customers()
Customers.customer_id
`````

if your variable is different every instance you have to make it as a Static variable and if it is same everyinstance you can make it instance variable. Remember that you can create your instance and static variable as a private variable as well.

But for the Static variable specifically.

`````python
class Customers:

    __customer_id = 0

    def __init__(self):
        self.name = ""
        self.customer_id = Customers.__customer_id
        Customers.__customer_id += 1
    
    @staticmethod # These kind of methods are called static methods. This decorator is used to show them.
    def get_customer_id():  # Without self also you can access this because you are using the name of the class in the return statement. And we are not returning any kind of data which is present inside the constructor.
        return Customers.__customer_id
`````

`````python
c1 = Customers()
`````

`````python
Customers.get_customer_id()
`````

In the above code name and gender are simple variables because they can be in one word. But address is a combination of multiple words which includes City, State, Country with your local address and pin code. for that we can create another class.
