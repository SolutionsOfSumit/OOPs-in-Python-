# Modular Coding in Python
in this type of coding we have to create a file as constructor named *__init__.py* in the root folder of our project which is how real world projects works.

Lets say your file structure looks something like this

```bash
project_File/
|    project/
|    |   |-- __init__.py
|    |   |--project_main.py
|    main.py
```

and there is a class in the `project.main.py` file that you need in `main.py` file there for you will use the code.


```python
from project.project_main import class_name

instance = class_name()

instance.method1()
```

# __name__ = "__main__"
It is a python feature that controls whether a file is being run directly or imported as a module.

#### Why is it used?
Suppose you have to a class

````python
class Car:
    def __init__(self, color):
        self.color = color

    def drive(self):
        print(f"{self.color} car is driving.")
````

You want to test teh class inside the same file.

````python
class Car:
    def __init__(self, color):
        self.color = color

    def drive(self):
        print(f"{self.color} car is driving.")

if __name__ == "__main__":
    car = Car("Red")
    car.drive()
````

If the question comes in your mind that you can write the same code as 
```python 
class Car:
    def __init__(self, color):
        self.color = color

    def drive(self):
        print(f"{self.color} car is driving.")

car = Car("Red")
car.drive()
```
the code will run perfectly fine for the single file and if you will import the file with the code 
```python 
from car import Car

my_car = Car("Blue")
my_car.drive()
```
it will give output like 
```
Red car is driving.
Blue car is driving.
```

but if you use `if __name__"__main__"` and import it somewhere else with code
```python 
from car import Car

my_car = Car("Blue")
my_car.drive()
```
it will print
```
Blue car is driving
```
That means the if code works if you run the same file without importing just for a test.
